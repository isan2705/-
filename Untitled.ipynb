```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. 读取数据
df1 = pd.read_csv('ReportCard1.txt', delim_whitespace=True)
df2 = pd.read_csv('ReportCard2.txt', delim_whitespace=True)

# 2. 合并数据
df = pd.merge(df1, df2, on='xh', how='outer')
df
```

```python
# 3. 计算总成绩和平均成绩
df['total'] = df[['chi', 'math', 'fore', 'phy', 'che', 'geo', 'his']].sum(axis=1)
df['average'] = df['total'] / 7
df['total']
df['average']
```

```python
df = df.sort_values(by='total', ascending=False)
df
```

```python
# 5. 按性别分别计算各门课程的平均成绩
male_avg = df[df['sex'] == 1.00][['chi', 'math', 'fore', 'phy', 'che', 'geo', 'his']].mean()
female_avg = df[df['sex'] == 2.00][['chi', 'math', 'fore', 'phy', 'che', 'geo', 'his']].mean()
male_avg
female_avg
```

```python
def grade分类(average):
    if average >= 90:
        return '优'
    elif average >= 80:
        return '良'
    elif average >= 70:
        return '中'
    elif average >= 60:
        return '及格'
    else:
        return '不及格'

df['grade'] = df['average'].apply(grade分类)
df['grade']
```

```python
male_grade_count = df[df['sex'] == 1.00]['grade'].value_counts().sort_index()
female_grade_count = df[df['sex'] == 2.00]['grade'].value_counts().sort_index()
male_grade_count
female_grade_count
```

```python
df = pd.get_dummies(df, columns=['sex'])
df
```

```python
# 9. 绘制总成绩的直方图
plt.figure(figsize=(10, 6))
plt.hist(df['total'], bins=10, edgecolor='k')
plt.title('总成绩直方图')
plt.xlabel('总成绩')
plt.ylabel('频数')
plt.show()
```

```python
# 10. 绘制平均成绩的优、良、中、及格和不及格的饼图
grade_counts = df['grade'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(grade_counts, labels=grade_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('平均成绩等级分布饼图')
plt.show()
```

```python
# 11. 绘制总成绩和数学(math)成绩的散点图
plt.figure(figsize=(10, 6))
plt.scatter(df['total'], df['math'], alpha=0.7)
plt.title('总成绩与数学成绩散点图')
plt.xlabel('总成绩')
plt.ylabel('数学成绩')
plt.show()
```
